## 카테고리 점수화 로직


## 카테고리 점수화 함수
이 셀은 제목, 본문, 부서, 이미지 OCR 텍스트에 포함된 키워드를 세고, 필드별 가중치를 적용해 카테고리 점수를 계산합니다.

현재 가중치 구조는 다음 의도를 갖습니다.
- 제목: 공지의 핵심 주제가 직접 드러나므로 높은 가중치
- 본문: 세부 맥락을 보완하는 가중치
- 부서: 담당 부서를 통해 주제를 추정하는 보조 가중치
- 이미지 텍스트: 포스터형 공지에서 누락 정보를 보완하는 가중치

이 점수는 LLM 응답 이전 단계에서 질문과 관련 있는 공지를 먼저 좁히는 규칙 기반 전처리 값으로 사용합니다.


```python
import pandas as pd
File_title = "일반공지"
# 🔹 CSV 파일 로드
file_path = "일반공지+이미지텍스트_포함.csv"
df = pd.read_csv(file_path)

# 🔹 키워드 정의 (간략 버전)
category_dict = {
    "일반공지": 일반_공지,
    "장학공지": 장학_공지,
    "학사공지": 채용_공지
}
department_dict = {
    "일반공지": 부서_분류,
    "장학공지": 부서_분류,
    "학사공지": 부서_분류
}
# 가중치 설정
TITLE_WEIGHT = 3
CONTENT_WEIGHT = 2
DEPT_WEIGHT = 0
IMAGE_WEIGHT = 2

def calculate_category_scores_by_notice_type(df: pd.DataFrame) -> pd.DataFrame:

    def calculate_score(title, content, dept, image, category_keywords, department_keywords):
        title = str(title) if pd.notna(title) else ""
        content = str(content) if pd.notna(content) else ""
        dept = str(dept) if pd.notna(dept) else ""
        image = str(image) if pd.notna(image) else ""

        scores = {k: 0 for k in category_keywords.keys()}

        for cat, keywords in category_keywords.items():
            scores[cat] += (sum(title.count(k) for k in keywords) * TITLE_WEIGHT * 10) / (len(title) + 1)
            scores[cat] += (sum(content.count(k) for k in keywords) * CONTENT_WEIGHT * 10) / (len(content) + 1)
            scores[cat] += (sum(image.count(k) for k in keywords) * IMAGE_WEIGHT * 10) / (len(image) + 1)

        for cat, keywords in department_keywords.items():
            if cat in scores:
                scores[cat] += (sum(dept.count(k) for k in keywords) * DEPT_WEIGHT * 10) / (len(dept) + 1)

        if all(v == 0 for v in scores.values()):
            scores["기타"] = 1.0

        return scores

    # 점수 계산 및 열 추가
    result_df = df.copy()
    for idx, row in result_df.iterrows():
        noti_type = row.get("공지분류", "").strip()
        category_keywords = category_dict.get(noti_type, {})
        department_keywords = department_dict.get(noti_type, {})

        title = row.get("공지 제목", "")
        content = row.get("본문 내용", "")
        dept = row.get("부서", "")
        image = row.get("이미지 텍스트", "")

        scores = calculate_score(title, content, dept, image, category_keywords, department_keywords)

        for cat, score in scores.items():
            colname = f"{noti_type}_{cat}"
            if colname not in result_df.columns:
                result_df[colname] = 0.0
            result_df.at[idx, colname] = score

    return result_df


#step6. 카테고리분류 지정
print("---> step6 시작")
try:
    step6_df = calculate_category_scores_by_notice_type(df)

    #파일 저장
    file_name = f"①_{File_title}+카테고리분류({TITLE_WEIGHT}-{CONTENT_WEIGHT}-{DEPT_WEIGHT}-{IMAGE_WEIGHT}).csv"
    step6_df.to_csv(file_name, index=False, encoding='utf-8-sig')
    print(f"---> step6 완료 : 저장파일명 = {file_name}")
except Exception as e:
    print("---> step6 실패")
    print(f"에러 내용: {e}")


# 🔹 최종 데이터프레임 출력
step6_df

```

```python
import pandas as pd

# 🔹 파일 로드 (사용할  CSV파일 이름 입력)
file_path = "통합공지_(마감일추출포함).csv"
df = pd.read_csv(file_path)
noti_case = '일반'

# 🔹 카테고리 그룹 딕셔너리
if noti_case == '일반':
    category_keywords = 일반_공지
    department_keywords = 부서_분류
elif noti_case == '장학':
    category_keywords = 장학_공지
    department_keywords = 부서_분류
elif noti_case == '채용':
    category_keywords = 채용_공지
    department_keywords = 부서_분류
elif noti_case == '학사':
    category_keywords = 학사_공지
    department_keywords = 부서_분류

# 🔹 가중치 설정
TITLE_WEIGHT = 3     # 공지 제목
CONTENT_WEIGHT = 2   # 본문 내용
DEPT_WEIGHT = 0      # 부서 이름
IMAGE_WEIGHT = 2     # 이미지 텍스트

# 🔹 카테고리 점수 환산 함수 (문자열 길이 기반)
def calculate_category_scores(title, content, dept, image_text):
    # NaN 처리
    title = str(title) if pd.notna(title) else ""
    content = str(content) if pd.notna(content) else ""
    dept = str(dept) if pd.notna(dept) else ""
    image_text = str(image_text) if pd.notna(image_text) else ""

    # 카테고리별 점수 초기화
    category_scores = {category: 0 for category in category_keywords.keys()}

    # 🔸 공지 제목 점수
    for category, keywords in category_keywords.items():
        title_count = sum(title.count(keyword) for keyword in keywords)
        category_scores[category] += ( title_count * TITLE_WEIGHT*10 ) / (len(title) + 1)

    # 🔸 본문 내용 점수
    for category, keywords in category_keywords.items():
        content_count = sum(content.count(keyword) for keyword in keywords)
        category_scores[category] += ( content_count * CONTENT_WEIGHT*10 ) / (len(content) + 1)

    # 🔸 부서 이름 점수
    for category, keywords in department_keywords.items():
        dept_count = sum(dept.count(keyword) for keyword in keywords)
        if category in category_scores:
            category_scores[category] += ( dept_count * DEPT_WEIGHT*10 ) / (len(dept) + 1)

    # 🔸 이미지 텍스트 점수
    for category, keywords in category_keywords.items():
        image_count = sum(image_text.count(keyword) for keyword in keywords)
        category_scores[category] += ( image_count * IMAGE_WEIGHT*10 ) / (len(image_text) + 1)

    # 🔹 모든 점수가 0일 경우 "기타" 카테고리 추가
    if all(score == 0 for score in category_scores.values()):
        category_scores["기타"] = 1.0

    return category_scores

# 🔹 데이터프레임에 카테고리 점수 추가
category_columns = list(category_keywords.keys()) + ["기타"]
for col in category_columns:
    if col not in df.columns:
        df[col] = 0.0

# 🔹 각 공지사항에 대해 점수 계산
for idx, row in df.iterrows():
    title = row.get("공지 제목", "")
    content = row.get("본문 내용", "")
    dept = row.get("부서", "")
    image_text = row.get("이미지 텍스트", "")

    scores = calculate_category_scores(title, content, dept, image_text)
    for category, score in scores.items():
        df.at[idx, category] = score

# 🔹 결과 저장
output_file = f"①_{noti_case}+마감일v1_카테고리_점수환산({TITLE_WEIGHT}-{CONTENT_WEIGHT}-{DEPT_WEIGHT}-{IMAGE_WEIGHT}).csv"
df.to_csv(output_file, index=False, encoding='utf-8-sig')
print(f"✅ 카테고리 점수가 포함된 데이터프레임이 저장되었습니다: {output_file}")

# 🔹 최종 데이터프레임 출력
df
```

## 분류 검증 코드


### 분류 검증용 상세 결과 생성
카테고리 점수뿐 아니라 어떤 키워드가 어느 필드에서 매칭됐는지 확인하기 위한 검증용 출력입니다. 분류 결과가 이상할 때 원인을 추적하고 키워드 사전을 수정하는 데 사용됩니다.



```python
import pandas as pd

# 🔹 파일 로드 (사용할  CSV파일 이름 입력)
file_path = "일반공지+이미지텍스트_포함.csv"
df = pd.read_csv(file_path)
noti_case = '일반'

# 🔹 키워드 사전 정의
# 아래는 예시이며 실제 실행 시에는 해당 변수들이 정의되어 있어야 합니다.
# from 외부_모듈_또는_사전_정의 import 일반_공지, 장학_공지, 채용_공지, 학사_공지, 부서_분류

if noti_case == '일반':
    category_keywords = 일반_공지
    department_keywords = 부서_분류
elif noti_case == '장학':
    category_keywords = 장학_공지
    department_keywords = 부서_분류
elif noti_case == '채용':
    category_keywords = 채용_공지
    department_keywords = 부서_분류
elif noti_case == '학사':
    category_keywords = 학사_공지
    department_keywords = 부서_분류

# 🔹 가중치 설정
TITLE_WEIGHT = 3
CONTENT_WEIGHT = 2
DEPT_WEIGHT = 0
IMAGE_WEIGHT = 2

# 🔹 점수 계산 함수
def calculate_category_scores(title, content, dept, image_text):
    title = str(title) if pd.notna(title) else ""
    content = str(content) if pd.notna(content) else ""
    dept = str(dept) if pd.notna(dept) else ""
    image_text = str(image_text) if pd.notna(image_text) else ""

    total_length = len(title) + len(content) + 1  # 0 방지
    category_scores = {cat: 0 for cat in category_keywords}
    category_weights = {}

    for cat, keywords in category_keywords.items():
        t_score = (sum(title.count(k) for k in keywords) * TITLE_WEIGHT  *10 ) / (len(title) + 1)
        c_score = (sum(content.count(k) for k in keywords) * CONTENT_WEIGHT * 10 ) / (len(content) + 1)
        i_score = (sum(image_text.count(k) for k in keywords) * IMAGE_WEIGHT *10 ) / (len(image_text) + 1)


        category_scores[cat] += t_score + c_score + i_score
        category_weights[cat] = f"제목{t_score}/본문{c_score}/이미지{i_score}"

    for cat, keywords in department_keywords.items():
        d_score = sum(dept.count(k) for k in keywords) * DEPT_WEIGHT
        if cat in category_scores:
            category_scores[cat] += d_score
            category_weights[cat] += f"/부서{d_score}"

    final_scores = {}
    final_weights = {}
    if all(score == 0 for score in category_scores.values()):
        final_scores["기타"] = 1.0
    else:
        for cat, score in category_scores.items():
            if score > 0:
                final_weights[cat] = category_weights[cat]

    return final_scores, final_weights

# 🔹 결과 열 초기화
category_columns = list(category_keywords.keys()) + ["기타"]
for cat in category_columns:
    df[cat] = 0.0
    df[f"{cat}_가중치"] = ""

# 🔹 점수 계산 적용
for idx, row in df.iterrows():
    title = row.get("공지 제목", "")
    content = row.get("본문 내용", "")
    dept = row.get("부서", "")
    image_text = row.get("이미지 텍스트", "")

    scores, weights = calculate_category_scores(title, content, dept, image_text)
    for cat, score in scores.items():
        df.at[idx, cat] = score
    for cat, detail in weights.items():
        df.at[idx, f"{cat}_가중치"] = detail

# 🔹 결과 저장
output_file = f"①_가중치_{noti_case}_카테고리_점수환산({TITLE_WEIGHT}-{CONTENT_WEIGHT}-{DEPT_WEIGHT}-{IMAGE_WEIGHT})_with_detail.csv"
df.to_csv(output_file, index=False, encoding='utf-8-sig')
print(f"✅ 카테고리 점수가 포함된 파일이 저장되었습니다: {output_file}")

# 🔹 결과 출력
df
```

## 알고리즘 단위 테스트

```python
# 🔹 카테고리 점수 환산 함수 (가중치 포함)
def calculate_category_scores(title, content="", dept=""):
    # NaN 처리
    title = str(title) if pd.notna(title) else ""
    content = str(content) if pd.notna(content) else ""
    dept = str(dept) if pd.notna(dept) else ""

    # 카테고리별 점수 초기화
    category_scores = {category: 0 for category in 일반_공지.keys()}
    total_score = 0

    # 🔸 공지 제목 점수
    for category, keywords in 일반_공지.items():
        title_count = sum(title.count(keyword) for keyword in keywords)
        category_scores[category] += title_count * TITLE_WEIGHT
        total_score += title_count * TITLE_WEIGHT

    # 🔸 본문 내용 점수
    for category, keywords in 일반_공지.items():
        content_count = sum(content.count(keyword) for keyword in keywords)
        category_scores[category] += content_count * CONTENT_WEIGHT
        total_score += content_count * CONTENT_WEIGHT

    # 🔸 평균 점수 환산
    if total_score > 0:
        for category in category_scores:
            if category_scores[category] > 0:
                category_scores[category] = round(category_scores[category] / total_score, 4)

    return category_scores
# 🔹 테스트 함수
def test_category_classification(text):
    scores = calculate_category_scores(text)
    print("\n📌 테스트 입력:", text)
    for category, score in scores.items():
        print(f"{category}: {score}")

# 🔹 테스트 실행
test_category_classification("창업 경진대회 캡프")
test_category_classification("성적우수 장학금 신청 안내")
test_category_classification("신입생을 위한 진로설계 프로그램")
test_category_classification("졸업생을 위한 취업 박람회 개최")
```

### 형태소 분석 기반 키워드 후보 추출
`konlpy`의 Okt 형태소 분석기를 사용해 공지 제목에서 자주 등장하는 명사를 추출합니다. 이 결과는 카테고리 키워드 사전을 만들거나 보완할 때 참고할 수 있습니다.



```python
import pandas as pd
from konlpy.tag import Okt
from collections import Counter

# 파일 경로
file_path = "일반공지.csv"

# 데이터 로드
df = pd.read_csv(file_path)

# 형태소 분석기 초기화
okt = Okt()

# 형태소 빈도수 계산
all_morphemes = []

# "공지 제목" 열의 형태소 추출
for title in df["공지 제목"].fillna(""):
    # 형태소 분리
    morphemes = okt.nouns(title)
    # 특수문자, 숫자 제거 (1글자 이하도 제외)
    morphemes = [token for token in morphemes if len(token) > 1 and not token.isdigit()]
    all_morphemes.extend(morphemes)

# 형태소 빈도수 계산
morpheme_freq = Counter(all_morphemes)

# 데이터프레임으로 변환
freq_df = pd.DataFrame(morpheme_freq.items(), columns=["형태소", "빈도수"])
freq_df = freq_df.sort_values(by="빈도수", ascending=False).reset_index(drop=True)

# 결과 저장
freq_df.to_csv("일반_제목_형태소.csv", index=False, encoding='utf-8-sig')
print("✅ 형태소 빈도수 데이터프레임이 title_morpheme_frequency.csv로 저장되었습니다.")

# 결과 미리보기
print(freq_df.head(20))
```

### 이전 버전 일반공지 분류 저장
일반공지 OCR 포함 파일을 대상으로 카테고리 점수를 계산하고 저장하는 이전 방식입니다. 현재 구조와 중복될 수 있으므로, 재사용 시 입력 파일과 가중치 설정을 먼저 확인해야 합니다.



```python
import pandas as pd

# 🔹 파일 로드
file_path = "일반공지+이미지텍스트_포함.csv"
df = pd.read_csv(file_path)
noti_case = '일반'

# 🔹 카테고리 그룹 딕셔너리
if noti_case == '일반':
    category_keywords = 일반_공지
    department_keywords = 부서_분류
elif noti_case == '장학':
    category_keywords = 장학_공지
    department_keywords = 부서_분류
elif noti_case == '채용':
    category_keywords = 채용_공지
    department_keywords = 부서_분류
elif noti_case == '학사':
    category_keywords = 학사_공지
    department_keywords = 부서_분류

# 🔹 가중치 설정
TITLE_WEIGHT = 3     # 공지 제목
CONTENT_WEIGHT = 2   # 본문 내용
DEPT_WEIGHT = 0      # 부서 이름
IMAGE_WEIGHT = 2     # 이미지 텍스트

# 🔹 카테고리 점수 환산 함수 (가중치 포함)
def calculate_category_scores(title, content, dept, image_text):
    # NaN 처리
    title = str(title) if pd.notna(title) else ""
    content = str(content) if pd.notna(content) else ""
    dept = str(dept) if pd.notna(dept) else ""
    image_text = str(image_text) if pd.notna(image_text) else ""

    # 카테고리별 점수 초기화
    category_scores = {category: 0 for category in category_keywords.keys()}
    total_score = 0

    # 🔸 공지 제목 점수
    for category, keywords in category_keywords.items():
        title_count = sum(title.count(keyword) for keyword in keywords)
        category_scores[category] += title_count * TITLE_WEIGHT
        total_score += title_count * TITLE_WEIGHT

    # 🔸 본문 내용 점수
    for category, keywords in category_keywords.items():
        content_count = sum(content.count(keyword) for keyword in keywords)
        category_scores[category] += content_count * CONTENT_WEIGHT
        total_score += content_count * CONTENT_WEIGHT

    # 🔸 부서 이름 점수
    for category, keywords in department_keywords.items():
        dept_count = sum(dept.count(keyword) for keyword in keywords)
        if category in category_scores:
            category_scores[category] += dept_count * DEPT_WEIGHT
            total_score += dept_count * DEPT_WEIGHT

    # 🔸 이미지 텍스트 점수
    for category, keywords in category_keywords.items():
        image_count = sum(image_text.count(keyword) for keyword in keywords)
        category_scores[category] += image_count * IMAGE_WEIGHT
        total_score += image_count * IMAGE_WEIGHT

    # 🔸 평균 점수 환산
    if total_score > 0:
        for category in category_scores:
            if category_scores[category] > 0:
                category_scores[category] = round(category_scores[category] / total_score, 4)
    else:
        # 🔹 모든 점수가 0일 경우 "기타" 카테고리 추가
        category_scores["기타"] = 1.0

    return category_scores

# 🔹 데이터프레임에 카테고리 점수 추가
category_columns = list(category_keywords.keys()) + ["기타"]
for col in category_columns:
    if col not in df.columns:
        df[col] = 0.0

# 🔹 각 공지사항에 대해 점수 계산
for idx, row in df.iterrows():
    title = row.get("공지 제목", "")
    content = row.get("본문 내용", "")
    dept = row.get("부서", "")
    image_text = row.get("이미지 텍스트", "")

    scores = calculate_category_scores(title, content, dept, image_text)
    for category, score in scores.items():
        df.at[idx, category] = score

# 🔹 결과 저장
output_file = f"{noti_case}_카테고리_점수환산({TITLE_WEIGHT}-{CONTENT_WEIGHT}-{DEPT_WEIGHT}-{IMAGE_WEIGHT}).csv"
df.to_csv(output_file, index=False, encoding='utf-8-sig')
print(f"✅ 카테고리 점수가 포함된 데이터프레임이 저장되었습니다: {output_file}")

# 🔹 최종 데이터프레임 출력
df
```
